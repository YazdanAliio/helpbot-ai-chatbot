```python
# cube.py — smooth rotating ASCII cube
import math
import time
import os
import sys

# Screen size (tweak these)
WIDTH = 96
HEIGHT = 40

# Cube size and sampling resolution (higher `step` => smoother but slower)
CUBE_HALF = 1.0
STEP = 0.06  # try 0.05 for even smoother but slower

# Characters from dim to bright
GRADIENT = " .:-=+*#%@"

# Light direction (normalized)
LIGHT = (0.0, 1.0, -1.0)
ln = math.sqrt(LIGHT[0]**2 + LIGHT[1]**2 + LIGHT[2]**2)
LIGHT = (LIGHT[0]/ln, LIGHT[1]/ln, LIGHT[2]/ln)

# Projection parameters
DIST = 4.0      # distance of camera from origin
PROJ_SCALE = 20 # scale projection onto screen

# Ensure terminal supports ANSI; on Windows newer terminals it works
def move_cursor_home():
    # Move cursor to home (top-left) without clearing screen
    sys.stdout.write("\033[H")
    sys.stdout.flush()

# Print enough newlines once to create the drawing area
print("\n" * HEIGHT)

def rotate_point(x, y, z, ax, ay, az):
    # Rotate point by ax, ay, az (radians) around x, y, z axes
    # Rx
    ca, sa = math.cos(ax), math.sin(ax)
    y, z = y * ca - z * sa, y * sa + z * ca
    # Ry
    cb, sb = math.cos(ay), math.sin(ay)
    x, z = x * cb + z * sb, -x * sb + z * cb
    # Rz
    cc, sc = math.cos(az), math.sin(az)
    x, y = x * cc - y * sc, x * sc + y * cc
    return x, y, z

def project(x, y, z):
    # simple perspective projection
    z_cam = z + DIST
    if z_cam == 0:
        z_cam = 1e-6
    px = int(WIDTH/2  + (x / z_cam) * PROJ_SCALE)
    py = int(HEIGHT/2 - (y / z_cam) * (PROJ_SCALE * 0.5))
    return px, py, z_cam

# define faces as (constant axis, value, u_axis, v_axis, normal)
# axis: 0=x,1=y,2=z. u_axis/v_axis indicate which axes are used for face grid
faces = [
    (0,  CUBE_HALF, 1, 2, (1, 0, 0)),  # +X face
    (0, -CUBE_HALF, 1, 2, (-1, 0, 0)), # -X face
    (1,  CUBE_HALF, 0, 2, (0, 1, 0)),  # +Y face
    (1, -CUBE_HALF, 0, 2, (0, -1, 0)), # -Y face
    (2,  CUBE_HALF, 0, 1, (0, 0, 1)),  # +Z face
    (2, -CUBE_HALF, 0, 1, (0, 0, -1)), # -Z face
]

def render(ax, ay, az):
    # z-buffer and char buffer
    zbuf = [0.0] * (WIDTH * HEIGHT)
    buf  = [' '] * (WIDTH * HEIGHT)

    # For each face sample the surface
    for face in faces:
        axis, val, u_axis, v_axis, normal_local = face
        # iterate u and v across the face
        u = -CUBE_HALF
        while u <= CUBE_HALF:
            v = -CUBE_HALF
            while v <= CUBE_HALF:
                # build point in object coords
                p = [0.0, 0.0, 0.0]
                p[axis] = val
                p[u_axis] = u
                p[v_axis] = v

                # rotated point
                rx, ry, rz = rotate_point(p[0], p[1], p[2], ax, ay, az)

                # compute normal rotated
                nx, ny, nz = rotate_point(normal_local[0], normal_local[1], normal_local[2], ax, ay, az)

                # lighting: dot(normal, light)
                ndotl = nx*LIGHT[0] + ny*LIGHT[1] + nz*LIGHT[2]
                if ndotl <= 0:
                    # backface / not lit; still could be visible, but dim
                    brightness = 0.0
                else:
                    brightness = ndotl

                # project to screen
                px, py, zcam = project(rx, ry, rz)

                if 0 <= px < WIDTH and 0 <= py < HEIGHT:
                    idx = px + py * WIDTH
                    # use inverse z for depth comparison (closer -> larger value)
                    depth = 1.0 / zcam
                    if depth > zbuf[idx]:
                        zbuf[idx] = depth
                        # map brightness to character index
                        char_idx = int(brightness * (len(GRADIENT) - 1))
                        # clamp
                        if char_idx < 0:
                            char_idx = 0
                        if char_idx >= len(GRADIENT):
                            char_idx = len(GRADIENT) - 1
                        buf[idx] = GRADIENT[char_idx]
                v += STEP
            u += STEP

    # convert buffer to lines
    lines = []
    for row in range(HEIGHT):
        start = row * WIDTH
        lines.append("".join(buf[start:start+WIDTH]))
    return "\n".join(lines)

def main():
    ax = ay = az = 0.0
    # Animation loop
    try:
        while True:
            frame = render(ax, ay, az)
            move_cursor_home()
            sys.stdout.write(frame)
            sys.stdout.flush()
            # update rotation
            ax += 0.03   # X-axis rotation speed
            ay += 0.04   # Y-axis rotation speed
            az += 0.02   # Z-axis rotation speed
            time.sleep(0.025)  # tweak for smoothness / CPU
    except KeyboardInterrupt:
        # move cursor below drawing before exit
        print("\n" * 2)
        print("Exiting.")

if __name__ == "__main__":
    main()
```

```python
import random
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time


data = [random.randint(10, 100) for _ in range(25)]

def dynamic_bubble_sort(data, delay=0.05):
    n = len(data)

    for i in range(n):
        for j in range(n - i - 1):

            clear_output(wait=True)
            plt.figure(figsize=(10,4))

            colors = ['blue'] * n
            colors[j] = 'red'
            colors[j+1] = 'red'

            plt.bar(range(n), data, color=colors)
            plt.title("Dynamic Bubble Sort")
            plt.ylim(0, max(data) + 10)
            plt.xlabel("Index")
            plt.ylabel("Value")

            plt.show()
            time.sleep(delay)

            if data[j] > data[j+1]:
                data[j], data[j+1] = data[j+1], data[j]


dynamic_bubble_sort(data)
```

```python
# Colab-ready Mandelbrot zoom (correct escape-time algorithm)
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time

def mandelbrot(w, h, x_center, y_center, zoom, max_iter=500):
    # compute window in complex plane
    x_width = 4.0 / zoom
    y_height = (h / w) * x_width
    x = np.linspace(x_center - x_width/2, x_center + x_width/2, w)
    y = np.linspace(y_center - y_height/2, y_center + y_height/2, h)
    X, Y = np.meshgrid(x, y)
    C = X + 1j * Y
    Z = np.zeros_like(C)
    M = np.full(C.shape, max_iter, dtype=float)   # store escape iteration (default = max_iter)

    for i in range(max_iter):
        mask = np.abs(Z) <= 2.0
        Z[mask] = Z[mask] * Z[mask] + C[mask]
        escaped = (np.abs(Z) > 2.0) & (M == max_iter)  # newly escaped this iteration
        M[escaped] = i

    # for nicer coloring, return normalized continuous iteration counts
    # smooth coloring: n + 1 - log(log|z|)/log 2
    with np.errstate(divide='ignore', invalid='ignore'):
        zn = Z
        nu = np.zeros_like(M)
        mask_esc = M < max_iter
        nu[mask_esc] = M[mask_esc] + 1 - np.log(np.log(np.abs(zn[mask_esc])))/np.log(2)
        nu[~mask_esc] = max_iter
    return nu

# Parameters (pick a center below)
x_center = -0.743643887037151   # Seahorse Valley (example)
y_center = 0.13182590420533
zoom = 2

plt.figure(figsize=(6,6))

for step in range(30):          # 30 frames; adjust as you like
    clear_output(wait=True)
    zoom *= 2.0                # doubling zoom each frame
    img = mandelbrot(800, 800, x_center, y_center, zoom, max_iter=600)
    plt.imshow(img, cmap='inferno', extent=[0,1,0,1], interpolation='bilinear')
    plt.title(f"Mandelbrot zoom — center=({x_center:.12f}, {y_center:.12f})  zoom={zoom:.2e}")
    plt.axis('off')
    plt.show()
    time.sleep(0.2)
```

```python
# Black Hole Gravitational Lensing Simulator (Colab-ready)
# Vectorized thin-lens approximation with animation of Einstein radius.
# Uses bilinear sampling of a generated starfield background.

import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML
from PIL import Image, ImageDraw

# ---------- Parameters ----------
W = 800       # output image width
H = 800       # output image height
n_stars = 2000  # number of background stars
seed = 42

# shadow / photon-sphere parameters (in pixels)
shadow_radius_px = 40.0  # radius of black hole "shadow" in pixels at base mass
shadow_scale = 1.0       # scales with mass

# animation frames
frames = 60

# Whether to save an mp4 (requires ffmpeg available in the environment)
SAVE_MP4 = False
MP4_FILENAME = "blackhole_lens.mp4"

np.random.seed(seed)

# ---------- Utility: generate starfield background ----------
def make_starfield(w, h, n, glow=True):
    # Create blank black PIL image
    img = Image.new("RGB", (w, h), color=(0,0,0))
    draw = ImageDraw.Draw(img)
    # random star positions, brightness, sizes
    xs = np.random.randint(0, w, size=n)
    ys = np.random.randint(0, h, size=n)
    mags = np.random.rand(n)**2  # skew to produce many faint stars, few bright
    for x, y, m in zip(xs, ys, mags):
        r = 1 if m < 0.7 else (2 if m < 0.95 else 3)
        brightness = int(120 + 135 * m)  # [120..255]
        # draw small ellipse for star
        draw.ellipse((x-r, y-r, x+r, y+r), fill=(brightness, brightness, brightness))
        if glow and r>1:
            # soft glow by drawing larger translucent ellipses
            glow_col = (int(brightness*0.4), int(brightness*0.5), int(brightness*0.9))
            draw.ellipse((x-3*r, y-3*r, x+3*r, y+3*r), fill=None, outline=None)
    return np.array(img).astype(np.float32) / 255.0

# background sky (you can replace this with an image you upload)
background = make_starfield(W, H, n_stars)

# ---------- Bilinear sampler ----------
def bilinear_sample(img, xs, ys):
    """
    img: H x W x C
    xs, ys: arrays of shape (H, W) with float coordinates (x in [0..W-1], y in [0..H-1])
    returns sampled image H x W x C (values in same dtype as img)
    """
    H, W, C = img.shape
    x0 = np.floor(xs).astype(int)
    x1 = x0 + 1
    y0 = np.floor(ys).astype(int)
    y1 = y0 + 1

    x0 = np.clip(x0, 0, W-1)
    x1 = np.clip(x1, 0, W-1)
    y0 = np.clip(y0, 0, H-1)
    y1 = np.clip(y1, 0, H-1)

    Ia = img[y0, x0]
    Ib = img[y1, x0]
    Ic = img[y0, x1]
    Id = img[y1, x1]

    wa = (x1 - xs) * (y1 - ys)
    wb = (x1 - xs) * (ys - y0)
    wc = (xs - x0) * (y1 - ys)
    wd = (xs - x0) * (ys - y0)

    wa = wa[..., None]
    wb = wb[..., None]
    wc = wc[..., None]
    wd = wd[..., None]

    return Ia*wa + Ib*wb + Ic*wc + Id*wd

# ---------- Lensing mapping (thin-lens, complex notation) ----------
# We'll model mapping: beta = theta - (theta_E^2 * theta / |theta|^2)
# where theta is observed angle vector (here pixel coordinates relative to center)
# theta_E controls lens strength (Einstein radius)
# This simple formula produces rings/arcs.

# generate pixel coordinate grids (theta)
xs = np.linspace(-1.0, 1.0, W)
ys = np.linspace(-1.0, 1.0, H)
X, Y = np.meshgrid(xs, ys)
theta_x = X.copy()
theta_y = Y.copy()
theta_r2 = theta_x**2 + theta_y**2
epsilon = 1e-8

# we'll map these normalized coords back to pixel coords when sampling background
def lens_map(theta_E_norm):
    """
    theta_E_norm: Einstein radius in the same normalized units as X,Y (i.e., fraction of half-image)
    returns source-plane coords (sx, sy) in normalized [-1,1] coordinates
    """
    # avoid division by zero at center
    denom = theta_r2 + epsilon
    fx = theta_E_norm**2 * theta_x / denom
    fy = theta_E_norm**2 * theta_y / denom
    beta_x = theta_x - fx
    beta_y = theta_y - fy
    return beta_x, beta_y

# convert normalized coords [-1,1] to pixel coords for sampling
def norm_to_pixels(nx, ny):
    # map normalized -1..1 to pixel coordinates 0..W-1 / 0..H-1
    px = ((nx + 1) / 2.0) * (W - 1)
    py = ((ny + 1) / 2.0) * (H - 1)
    return px, py

# ---------- Render a single frame ----------
def render_frame(theta_E_norm, shadow_px):
    # compute mapped source positions
    bx, by = lens_map(theta_E_norm)
    sx, sy = norm_to_pixels(bx, by)

    # sample background
    sampled = bilinear_sample(background, sx, sy)

    # compute shadow mask (circular dark region at center)
    cx = W // 2
    cy = H // 2
    y_idx, x_idx = np.indices((H, W))
    rr = np.sqrt((x_idx - cx)**2 + (y_idx - cy)**2)
    shadow_mask = rr <= shadow_px

    # create vignette/relativistic brightening near inner edge (simple artistic boost)
    ring_boost = np.exp(-((rr - shadow_px) / (shadow_px*0.6))**2)
    ring_boost = np.clip((1.0 - ring_boost)*2.0, 0.0, 2.0)[..., None]

    # darken inside shadow and optionally add faint accretion glow
    accretion = np.zeros_like(sampled)
    accretion[:,:,0] = 1.0  # subtle color in accretion disk (unused now)
    out = sampled * 1.0
    out[shadow_mask] = 0.0

    # add lensing-caused bright arcs near shadow edge (artistic)
    # compute normalized distance from shadow (0 at edge)
    dist_from_edge = (rr - shadow_px) / (shadow_px*0.8)
    arc_mask = (dist_from_edge >= 0) & (dist_from_edge < 1.2)
    # boost brightness for arc pixels (simulate light warping)
    out[arc_mask] = np.clip(out[arc_mask] * (1.0 + (1.2 - dist_from_edge[arc_mask])[:,None]*0.8), 0.0, 1.0)

    # subtle global tone mapping / gamma
    out = np.clip(out, 0.0, 1.0)
    out = out**0.9

    return out

# ---------- Prepare animation ----------
fig = plt.figure(figsize=(6,6))
ax = fig.add_subplot(111)
ax.axis('off')
im = ax.imshow(np.zeros((H, W, 3)), origin='upper', interpolation='bilinear')

def init():
    im.set_data(np.zeros((H, W, 3)))
    return (im,)

def animate(i):
    # animate theta_E from small to large and back for a nice visual
    t = i / frames  # 0..1
    # make lens strength pulse
    theta_E_norm = 0.02 + 0.12 * (0.5 - 0.5*np.cos(2*np.pi * t))  # normalized Einstein radius
    # convert normalized to shadow in pixels (simple proportional rule)
    shadow_px = shadow_radius_px * shadow_scale * (1.0 + 1.5 * (theta_E_norm - 0.02)/0.12)
    frame = render_frame(theta_E_norm, shadow_px)
    im.set_data(frame)
    ax.set_title(f"Gravitational Lensing — Einstein radius ~ {theta_E_norm:.3f}", color='white', pad=6)
    return (im,)

anim = animation.FuncAnimation(fig, animate, init_func=init, frames=frames, interval=60, blit=True)

# display in-notebook
plt.close(fig)  # prevent static image below the player
HTML(anim.to_html5_video())  # Colab will render the video player

# Optionally save to MP4 (uncomment if you want to save; requires ffmpeg)
if SAVE_MP4:
    print("Saving MP4 (this may take a moment)...")
    anim.save(MP4_FILENAME, fps=20, dpi=150, codec='libx264')
    print("Saved to", MP4_FILENAME)
```

```python
import random, time, os
from IPython.display import clear_output

cols = 80
rows = 24
chars = "01あいうえおアイウエオカキクケコ0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ"
drops = [0] * cols

try:
    while True:
        # build screen
        screen = []
        for r in range(rows):
            line = []
            for c in range(cols):
                # chance to draw a bright head
                if drops[c] == r:
                    line.append('\033[92m' + random.choice(chars) + '\033[0m')
                # trailing characters
                elif drops[c] > r and drops[c] - r < 6:
                    line.append('\033[32m' + random.choice(chars) + '\033[0m')
                else:
                    line.append(' ')
            screen.append(''.join(line))
        clear_output(wait=True)
        print('\n'.join(screen))
        # advance columns
        for i in range(cols):
            if random.random() > 0.975:
                drops[i] = 0
            else:
                drops[i] += 1
            # reset if fell off
            if drops[i] > rows + random.randint(0,10):
                drops[i] = 0
        time.sleep(0.06)
except KeyboardInterrupt:
    print("Stopped")
```
